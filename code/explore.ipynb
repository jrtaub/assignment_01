# Bill Splitter — Notebook

This notebook is the **notebook interface** for the Bill Splitter. Like the
console and Streamlit apps, it does no math of its own — it imports `bill.py`
and calls its functions to explore different tips.

Run each cell with **Shift+Enter** (choose the `/usr/local/bin/python` kernel the
first time), or use **Run All**.

```python
import pandas as pd

import bill
```

```python
# One example: a $50 bill, a 20% tip, split 4 ways.
subtotal, pct, people = 50.0, 20, 4
total = bill.grand_total(subtotal, pct)
per_person = bill.split_evenly(total, people)

print (f"Grand total: ${total:.2f}")
print (f"Each of {people} pays: ${per_person:.2f}")
per_person
```

```python
# Compare the per-person cost across a range of tip percentages.
percents = [10, 15, 18, 20, 25]
table = pd.DataFrame(
    {"per person": [bill.split_evenly(bill.grand_total(subtotal, p), people) for p in percents]},
    index=[f"{p}%" for p in percents],
)
table
```

```python
# A quick bar chart of the same data.
table.plot(kind="bar", legend=False, title="Per-person cost by tip %")
```
